```python
!pip install Sastrawi
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

paragraf = (
    "Pagi hari di halaman rumah terasa sangat tenang. "
    "Sinar matahari mulai masuk lewat celah-celah daun. "
    "Suara burung berkicau pelan menemani udara yang sejuk."
)

print("TEKS ASLI")
print(paragraf)
print()
```

```python
huruf_kecil = paragraf.lower()

# Tanda baca diganti spasi (bukan dihapus) agar kata tidak menyatu
bersih = re.sub(r"[^a-z0-9\s]", " ", huruf_kecil)
# Rapikan spasi berlebih
bersih = re.sub(r"\s+", " ", bersih).strip()

print("1. CASE FOLDING + PEMBERSIHAN TANDA BACA")
print("Huruf kecil :", huruf_kecil)
print("Bersih      :", bersih)
print()
```

```python
token = bersih.split()

print("2. TOKENISASI")
print("Jumlah token:", len(token))
print("Token       :", token)
print()
```

```python
stopwords = set(StopWordRemoverFactory().get_stop_words())
# Kata negasi tetap dipertahankan agar makna tidak hilang
stopwords = stopwords - {"tidak", "bukan", "jangan", "belum"}

token_penting = []
token_dihapus = []
for kata in token:
    if kata in stopwords:
        token_dihapus.append(kata)
    else:
        token_penting.append(kata)

print("3. PENGHAPUSAN STOPWORDS")
print("Token dihapus:", token_dihapus)
print("Jumlah token :", len(token), "->", len(token_penting))
print("Token tersisa:", token_penting)
print()
```

```python
stemmer = StemmerFactory().create_stemmer()
hasil_stem = [stemmer.stem(kata) for kata in token_penting]

print("4. STEMMING")
print("Sebelum:", token_penting)
print("Sesudah:", hasil_stem)
print()

# Kata yang berubah akibat stemming
perubahan = [(a, b) for a, b in zip(token_penting, hasil_stem) if a != b]
print("Kata yang berubah:")
for sebelum, sesudah in perubahan:
    print(f"  {sebelum} -> {sesudah}")
```
